```
import pandas as pd, seaborn as sns, matplotlib.pyplot as plt
sns.set_style('whitegrid')
df=pd.read_csv('../data/dataset_part_2.csv')
df.head()
```

```
sns.scatterplot(data=df, x='FlightNumber', y='LaunchSite', hue='Class')
plt.title('Flight Number vs Launch Site'); plt.show()
```

```
sns.scatterplot(data=df, x='PayloadMass', y='LaunchSite', hue='Class')
plt.title('Payload vs Launch Site'); plt.show()
```

```
site_succ = df.groupby('LaunchSite')['Class'].mean().sort_values(ascending=False)
site_succ.plot(kind='bar'); plt.title('Success rate by Site'); plt.ylabel('Success rate'); plt.show()
site_succ
```

```
orbit_succ = df.groupby('Orbit')['Class'].mean().sort_values(ascending=False)
orbit_succ.plot(kind='bar'); plt.title('Success rate by Orbit'); plt.ylabel('Success rate'); plt.show()
```

```
df['Year']=pd.to_datetime(df['Date']).dt.year
df.groupby('Year')['Class'].mean().plot(marker='o'); plt.title('Yearly success trend'); plt.show()
```

```
features = df.drop(columns=['Date','BoosterVersion','Outcome','LandingPad','Serial','Year'])
features = pd.get_dummies(features, columns=['Orbit','LaunchSite'])
features = features.astype(float)
features.to_csv('../data/dataset_part_3.csv', index=False)
print(features.shape)
```
