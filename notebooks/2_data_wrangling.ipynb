We label each flight with a binary `Class`:
- `1` = first stage successfully landed
- `0` = first stage failed or did not attempt landing

```
import pandas as pd, numpy as np
df = pd.read_csv('../data/dataset_part_1.csv')
df.head()
```

```
# Inspect missing values
print(df.isnull().sum())
```

```
# Frequency of launches at each site
print(df['LaunchSite'].value_counts())
```

```
# Frequency of each orbit
print(df['Orbit'].value_counts())
```

```
# Compute mean success rate
print('Overall success rate:', df['Class'].mean().round(3))
```
