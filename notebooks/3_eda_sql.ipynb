We load the SpaceX dataset into a SQLite database and answer the IBM lab's standard SQL questions.

```
import pandas as pd, sqlite3
con=sqlite3.connect(':memory:')
sp=pd.read_csv('../data/Spacex.csv')
sp.to_sql('SPACEXTBL', con, if_exists='replace', index=False)
print(sp.shape); sp.head()
```

### Q1. Unique launch sites

```
pd.read_sql('SELECT DISTINCT Launch_Site FROM SPACEXTBL', con)
```

### Q2. 5 records where launch sites begin with 'CCA'

```
pd.read_sql("SELECT * FROM SPACEXTBL WHERE Launch_Site LIKE 'CCA%' LIMIT 5", con)
```

### Q3. Total payload mass carried for NASA (CRS)

```
pd.read_sql("SELECT SUM(PAYLOAD_MASS__KG_) AS total FROM SPACEXTBL WHERE Customer='NASA (CRS)'", con)
```

### Q4. Average payload for booster version F9 v1.1

```
pd.read_sql("SELECT AVG(PAYLOAD_MASS__KG_) AS avg_payload FROM SPACEXTBL WHERE Booster_Version LIKE '%v1.1%'", con)
```

### Q5. Date of first successful ground-pad landing

```
pd.read_sql("SELECT MIN(DATE) AS first_success FROM SPACEXTBL WHERE \"Landing _Outcome\" LIKE '%True RTLS%'", con)
```

### Q6. Boosters that succeeded on drone ship with payload 4000<mass<6000

```
pd.read_sql("SELECT Booster_Version FROM SPACEXTBL WHERE \"Landing _Outcome\"='True ASDS' AND PAYLOAD_MASS__KG_ BETWEEN 4000 AND 6000", con)
```

### Q7. Total successful vs failed missions

```
pd.read_sql("SELECT Mission_Outcome, COUNT(*) FROM SPACEXTBL GROUP BY Mission_Outcome", con)
```

### Q8. Boosters that carried the maximum payload mass

```
pd.read_sql("SELECT Booster_Version FROM SPACEXTBL WHERE PAYLOAD_MASS__KG_=(SELECT MAX(PAYLOAD_MASS__KG_) FROM SPACEXTBL)", con)
```

### Q9. 2015 records: month, landing outcome, booster, launch site (failures on drone ship)

```
pd.read_sql("SELECT substr(DATE,6,2) AS month, \"Landing _Outcome\", Booster_Version, Launch_Site FROM SPACEXTBL WHERE substr(DATE,1,4)='2015' AND \"Landing _Outcome\" LIKE 'False ASDS%'", con)
```

### Q10. Rank landing outcomes between 2010-06-04 and 2017-03-20

```
pd.read_sql("SELECT \"Landing _Outcome\", COUNT(*) AS cnt FROM SPACEXTBL WHERE DATE BETWEEN '2010-06-04' AND '2017-03-20' GROUP BY \"Landing _Outcome\" ORDER BY cnt DESC", con)
```
