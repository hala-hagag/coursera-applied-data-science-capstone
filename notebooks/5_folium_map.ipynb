```
import pandas as pd, folium
from folium.plugins import MarkerCluster
from folium.features import DivIcon
df=pd.read_csv('../data/dataset_part_2.csv')
nasa=[29.559684888503615, -95.0830971930759]
m=folium.Map(location=nasa, zoom_start=4)
folium.Marker(nasa, icon=folium.Icon(color='red', icon='star'), popup='NASA JSC').add_to(m)
```

```
sites=df.groupby('LaunchSite')[['Latitude','Longitude']].mean().reset_index()
for _,r in sites.iterrows():
    folium.Circle([r['Latitude'],r['Longitude']], radius=1000, color='#000', fill=True).add_to(m)
    folium.Marker([r['Latitude'],r['Longitude']],
                  icon=DivIcon(icon_size=(20,20), icon_anchor=(0,0),
                               html=f'<div style="font-size:12;color:#0066cc"><b>{r["LaunchSite"]}</b></div>')).add_to(m)
m
```

```
cluster=MarkerCluster().add_to(m)
for _,row in df.iterrows():
    color='green' if row['Class']==1 else 'red'
    folium.Marker([row['Latitude'],row['Longitude']],
                  icon=folium.Icon(color=color),
                  popup=f"Flight {row['FlightNumber']} - {'Success' if row['Class']==1 else 'Failure'}").add_to(cluster)
m.save('../images/folium_map.html')
m
```
